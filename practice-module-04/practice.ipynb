# Problem 9

```python
input_value = input("Please enter a value: ")
lower_case = input_value.lower()
count = 0
for i in range(len(lower_case)):
    if(lower_case[i] == 'a' or lower_case[i] == 'e' or lower_case[i] == 'i' or lower_case[i] == 'o' or lower_case[i] == 'u'):
        count += 1
print(count)
```

# Problem 10

```python
list_of_marks = [90, 40, 75, 88, 35]
sum = 0
pass_count = 0
for i in range(len(list_of_marks)):
    if (list_of_marks[i] >= 40):
        pass_count+=1
        print(list_of_marks[i], "Pass")
    sum+=list_of_marks[i]
average = float(sum / len(list_of_marks))
print("Passed:", pass_count)
print("Average:", average)
```
