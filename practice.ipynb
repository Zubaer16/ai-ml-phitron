# Smart Light

# Count Even Numbers

```python
input_of_n = int(input("Enter a number:"))
input_of_numbers = input("Enter multiple numbers:")
split_numbers = input_of_numbers.split()
even_count = 0
for i in range(input_of_n):
    split_numbers[i] = int(split_numbers[i])
for i in range(input_of_n):
    if (split_numbers[i] % 2 == 0):
        even_count += 1
print (even_count)
```

# Smallest and Largest

```python
input_of_n = int(input("Enter a number:"))
input_of_numbers = input("Enter numbers:")
split_numbers = input_of_numbers.split()
min = 9999999999999
max = 0
for i in range(input_of_n):
    split_numbers[i] = int(split_numbers[i])
    if (split_numbers[i] < min):
            min = split_numbers[i]         
    if (split_numbers[i] > max):
            max = split_numbers[i]
    
print(min,max)
```

# Positive, Negative, Even and Odd

```python
input_of_n = int(input("Enter a number:"))
input_of_numbers = input("Enter numbers:")
split_numbers = input_of_numbers.split()
positive_count = 0 
negative_count = 0
even_count = 0
odd_count = 0
for i in range(input_of_n):
    split_numbers[i] = int(split_numbers[i])
    if( split_numbers[i] > 0):
        positive_count += 1
    if( split_numbers[i] < 0):
        negative_count += 1
    if( split_numbers[i] % 2 == 0):
        even_count += 1
    if( split_numbers[i] % 2 != 0):
        odd_count += 1
print("Positive:", positive_count)
print("Negative:", negative_count)
print("Even:", even_count)
print("Odd:", odd_count)
```

# Find the First Number Greater Than 100

```python
number = int(input("Enter a number:"))
numbers = input("Enter numbers:")
max = 100
split_numbers = numbers.split()
for i in range(number):
    split_numbers[i] = int(split_numbers[i])
    if (split_numbers[i] > max):
        max = split_numbers[i]
        break
print(max)
```

# Number Analysis

```python
number = int(input("Enter a number:"))
numbers = input("Enter numbers:")

split_numbers = numbers.split()
count = 0
for i in range(number):
    sum = 0
    if(int(split_numbers[i]) % 2 == 0):
        continue
    if( int(split_numbers[i]) % 2 != 0 ):
        for j in range(len(split_numbers[i])) :
            
            sum += int(split_numbers[i][j])
    if(sum > 20):
        break
    count += 1
    print(split_numbers[i], sum)
print(count)
```
