# Smart Light

```python
user_input = input ("Enter 2 brightness and threshold values:")
split_two_values = user_input.split()
for i in range(2) :
    split_two_values[i] = int(split_two_values[i])

if (split_two_values[0] > split_two_values[1]):
    print("On")
else:
    print("Off")
```

# Even or Odd

# Count Even Numbers

```python
input_of_n = int(input("Enter a number:"))
input_of_numbers = input("Enter multiple numbers:")
split_numbers = input_of_numbers.split()
even_count = 0
for i in range(input_of_n):
    split_numbers[i] = int(split_numbers[i])
for i in range(input_of_n):
    if (split_numbers[i] % 2 == 0):
        even_count += 1
print (even_count)
```

# Sum of Odd Numbers

```python
user_input = int(input("Enter a number:"))
odd_number_sum = 0
for i in range (1, user_input+ 1):
    if(i%2 !=0):
        odd_number_sum+=i

print(odd_number_sum)
```
