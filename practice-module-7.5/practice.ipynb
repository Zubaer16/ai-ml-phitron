# Create and Write to a File

```python
def create_file(filename, content):
    with open(filename,'w') as f:
        f.write(content)
create_file("notes.txt", "Hello, this is my first file.")

```

# Read Lines with Line Numbers

```python
with open ("data.txt", "r") as f:
    data = f.read()
values = data.split('\n')
for i in range(len(values)):
    if(i+1 == len(values)):
         break
    print(f"{i+1}: {values[i]}")
    
```

# Append Log Entries to a File

```python
def add_log(message):
    with open("log.txt","a") as f:
        f.write(message + "\n")
add_log("User logged in")
add_log("File uploaded")
```

# Read from a Specific File Pointer Position

```python
def read_from_position(filename, position):
    with open(filename,"r") as f:
        f.seek(position)
        return f.read()
read_from_position("hello.txt",7)
```

# Safe File Read with Full Exception Handling


```python
def safe_read(filename):
    try:
        with open(filename, "r") as f:
            data = f.read()
    except FileNotFoundError:
        print(f"Error: File not found.")
        
    else:
        print(data)
    finally:
        print("Operation complete")

safe_read("hello.txt")
    
```
