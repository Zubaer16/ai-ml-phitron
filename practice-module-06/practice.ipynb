# The Model Registry

```python
def model_info(name, layers, lr=0.01):
    print(f"Model: {name}, Layers: {layers}, LR: {lr}")
number = int(input())

for i in range(number):
    data = input()
    list_of_data = data.split()

    if(i == 0):
        model_info(list_of_data[1], list_of_data[2])
    if (i==1): 
        model_info(layers = list_of_data[2], name=list_of_data[1])
    if (i == 2):
        model_info(name=list_of_data[1], layers=list_of_data[2], lr=float(list_of_data[3]))
```

# GPU Can Eat Only Three Images (Hard)

```python
def batch_generator(data, batch_size):
    if batch_size < 1:
        return

    for i in range(0, len(data), batch_size):
        yield data[i:i + batch_size]


def batch_mean(batch):
    return sum(batch) / len(batch)


n, batch_size = map(int, input().split())
data = list(map(int, input().split()))

for batch in batch_generator(data, batch_size):
    print(f"{batch} mean={batch_mean(batch)}")
```
