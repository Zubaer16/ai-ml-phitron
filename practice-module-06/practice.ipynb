# The Homework Bot

```python
number = int(input() )
def check_pass(marks):
    if marks >= 40:
        return True
    else:
        return False

if check_pass(number) :
    print("Pass")
else:
    print("Fail")
```

# Three Training Runs

```python

def average(a,b,c):     
    return (a+b+c)/3
a,b,c = map(float,input().split())
print(average(a,b,c))
```
