# Problem 2

```python
data = [1, 2, 3, 4, 5, 6, 7, 8]
batch_size = 3
def batch_generator(list_of_numbers,batch_size):
    for i in range(0, len(list_of_numbers), batch_size):
        yield list_of_numbers[i:i+batch_size]
for batch in batch_generator(data, batch_size):
    print(batch)
```

# Problem 4

```python
input_of_base_exponent = list(map(int, input().split()))
def power(base, exponent=2):
    return base ** exponent

if len(input_of_base_exponent) == 1:
    result = power(input_of_base_exponent[0])
elif len(input_of_base_exponent) == 2:
    result = power(input_of_base_exponent[0], input_of_base_exponent[1])
print(result)
```
